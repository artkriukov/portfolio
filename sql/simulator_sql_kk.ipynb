Задача 1

Выведите всю информацию о товарах, цена которых не превышает 100 рублей. 

```python
SELECT product_id,
       name,
       price
FROM   products
WHERE  price <= 100
ORDER BY product_id
```

```python
SELECT user_id,
       order_id,
       time
FROM   user_actions
WHERE  time > '2022-09-06'
   and action = 'create_order'
ORDER BY order_id
```

Задача 4

Назначьте скидку 20% на все товары из таблицы products и отберите те, цена на которые с учётом скидки превышает 100 рублей. 

```python
SELECT product_id,
       name,
       price as old_price,
       price * 0.8 as new_price
FROM   products
WHERE  price * 0.8 > 100
ORDER BY product_id
```

Задача 5

Отберите из таблицы products все товары, названия которых либо начинаются со слова «чай», либо состоят из пяти символов.

```python
SELECT product_id,
       name
FROM   products
WHERE  split_part(name, ' ', 1) = 'чай'
    or length(name) = 5
ORDER BY product_id
```

Задача 6

```python
SELECT 
   product_id,
   name
FROM   products
WHERE  name like 'с%'
   and name not like '% %'
ORDER BY product_id
```

Задача 8

Выберите из таблицы products все чаи стоимостью больше 60 рублей и вычислите для них цену со скидкой 25%. Скидку в % менеджер попросил указать в отдельном столбце в формате текста, то есть вот так: «25%». Также необходимо любым способом избавиться от «чайного гриба»: вряд ли менеджер имел в виду и его, когда ставил нам задачу. 

```python
SELECT 
   product_id,
    name,
    price,
    '25%' as discount,
    price*0.75 as new_price
FROM   products
WHERE  name like '%чай %'
   and price > 60
ORDER BY product_id
```

Задача 16

Вычислите НДС каждого товара в таблице products и рассчитайте цену без учёта НДС. Однако теперь примите во внимание, что для товаров из списка налог составляет 10%. Для остальных товаров НДС тот же — 20%. 

```python
SELECT product_id,
       name,
       price,
       case 
          when name in ('сахар', 'сухарики', 'сушки', 'семечки', 'масло льняное', 
                      'виноград', 'масло оливковое', 'арбуз', 'батон', 'йогурт', 'сливки', 'гречка', 
                      'овсянка', 'макароны', 'баранина', 'апельсины', 'бублики', 'хлеб', 'горох', 
                      'сметана', 'рыба копченая', 'мука', 'шпроты', 'сосиски', 'свинина', 'рис', 
                      'масло кунжутное', 'сгущенка', 'ананас', 'говядина', 'соль', 'рыба вяленая', 
                      'масло подсолнечное', 'яблоки', 'груши', 'лепешка', 'молоко', 'курица', 
                      'лаваш', 'вафли', 'мандарины') 
          then round(price/110*10, 2)
        else round(price/120*20, 2) end as tax,
       case 
          when name in ('сахар', 'сухарики', 'сушки', 'семечки', 'масло льняное', 'виноград', 
                        'масло оливковое', 'арбуз', 'батон', 'йогурт', 'сливки', 'гречка', 'овсянка', 
                        'макароны', 'баранина', 'апельсины', 'бублики', 'хлеб', 'горох', 'сметана', 
                        'рыба копченая', 'мука', 'шпроты', 'сосиски', 'свинина', 'рис', 'масло кунжутное', 
                        'сгущенка', 'ананас', 'говядина', 'соль', 'рыба вяленая', 'масло подсолнечное', 
                        'яблоки', 'груши', 'лепешка', 'молоко', 'курица', 'лаваш', 'вафли', 'мандарины') 
          then round(price - price/110*10, 2)
        else round(price - price/120*20, 2) end as price_before_tax
FROM   products
ORDER BY price_before_tax desc, product_id
```

Задача 16

Разбейте пользователей из таблицы users на 4 возрастные группы: от 18 до 24 лет; от 25 до 29 лет; от 30 до 35 лет; не младше 36. Посчитайте число пользователей, попавших в каждую возрастную группу.
```python
SELECT case when date_part('year', age(now(), birth_date))::int >= 18 and
                 date_part('year', age(now(), birth_date))::int <= 24 then '18-24'
            when date_part('year', age(now(), birth_date))::int >= 25 and
                 date_part('year', age(now(), birth_date))::int <= 29 then '25-29'
            when date_part('year', age(now(), birth_date))::int >= 30 and
                 date_part('year', age(now(), birth_date))::int <= 35 then '30-35'
            when date_part('year', age(now(), birth_date))::int >= 36 then '36+' 
        end group_age,
       count(user_id) as users_count
FROM   users
WHERE  birth_date is not null
GROUP BY group_age
ORDER BY group_age
```

Задача 17

По данным из таблицы orders рассчитайте средний размер заказа по выходным и будням. Группу с выходными днями (суббота и воскресенье) назовите «weekend», а группу с будними днями (с понедельника по пятницу) — «weekdays»

```python
SELECT case when extract('dow' FROM creation_time) in (6, 0) then 'weekend' 
            when extract('dow' FROM   creation_time) < 6 then 'weekdays' 
        end as week_part, 
        round(avg(array_length(product_ids, 1)), 2) as avg_order_size
FROM   orders
GROUP BY week_part
ORDER BY avg_order_size
```

Задача 18

Для каждого пользователя в таблице user_actions посчитайте общее количество оформленных заказов и долю отменённых заказов. В результат включите только тех пользователей, которые оформили больше трёх заказов и у которых показатель cancel_rate составляет не менее 0.5. 

```python
SELECT user_id,
       round(count(distinct order_id) filter (WHERE action = 'cancel_order')::decimal / count(distinct order_id), 2) as cancel_rate,
       count(distinct order_id) as orders_count
FROM   user_actions
GROUP BY user_id 
having round(count(distinct order_id) filter (WHERE  action = 'cancel_order')::decimal / count(distinct order_id), 2) >= 0.5
   and count(distinct order_id) > 3
ORDER BY user_id
```

```python
SELECT date_part('isodow', time)::int as weekday_number,
       to_char(time, 'Dy') as weekday,
       count(order_id) filter (WHERE action = 'create_order') as created_orders,
       count(order_id) filter (WHERE action = 'cancel_order') as canceled_orders,
       count(order_id) filter (WHERE action = 'create_order') - count(order_id) filter (WHERE action = 'cancel_order') as actual_orders,
       round((count(order_id) filter (WHERE action = 'create_order') - count(order_id) filter (WHERE action = 'cancel_order'))::decimal / count(order_id) filter (WHERE action = 'create_order'),
             3) as success_rate
FROM   user_actions
WHERE  time >= '2022-08-24'
   and time < '2022-09-07'
GROUP BY weekday_number, weekday
ORDER BY weekday_number
```

Задача 14

На основе объединённой таблицы для каждого пользователя рассчитайте следующие показатели: общее число заказов — колонку назовите orders_count среднее количество товаров в заказе — avg_order_size суммарную стоимость всех покупок — sum_order_value среднюю стоимость заказа — avg_order_value минимальную стоимость заказа — min_order_value максимальную стоимость заказа — max_order_value 

```python
WITH valid_orders AS (
    SELECT DISTINCT 
        ua.user_id,
        ua.order_id
    FROM user_actions ua
    WHERE ua.order_id NOT IN (
        SELECT order_id 
        FROM user_actions 
        WHERE action = 'cancel_order'
    )
),

orders_info AS (
    SELECT 
        vo.user_id,
        vo.order_id,
        unnest(o.product_ids) AS product_id
    FROM valid_orders vo 
    JOIN orders o ON o.order_id = vo.order_id
),

order_stat AS (
    SELECT 
        oi.user_id,
        oi.order_id,
        COUNT(*) AS order_size,
        SUM(p.price) AS order_price
    FROM orders_info oi 
    JOIN products p ON p.product_id = oi.product_id
    GROUP BY oi.user_id, oi.order_id
)

SELECT 
    user_id,
    COUNT(order_id) AS orders_count,
    ROUND(AVG(order_size), 2) AS avg_order_size,
    SUM(order_price) AS sum_order_value,
    ROUND(AVG(order_price), 2) AS avg_order_value,
    MIN(order_price) AS min_order_value,
    MAX(order_price) AS max_order_value
FROM order_stat
GROUP BY user_id
ORDER BY user_id
LIMIT 1000
```

```python
WITH not_canceled_orders AS (
    SELECT 
        o.order_id,
        unnest(product_ids) AS product_id,
        creation_time,
        ua.action
    FROM orders o
    LEFT JOIN user_actions ua ON ua.order_id = o.order_id
    WHERE o.order_id NOT IN (
        SELECT order_id 
        FROM user_actions 
        WHERE action = 'cancel_order'
    )
),

order_price AS (
    SELECT 
        order_id,
        DATE(creation_time) AS date,
        SUM(p.price) AS order_price
    FROM not_canceled_orders nco 
    JOIN products p ON p.product_id = nco.product_id
    GROUP BY order_id, creation_time
)

SELECT 
    date,
    SUM(order_price)::DECIMAL AS revenue
FROM order_price
GROUP BY date
ORDER BY date
```

```python
WITH not_canceled AS (
    SELECT
        user_id,
        time,
        LAG(time) OVER (PARTITION BY user_id ORDER BY time) AS prev_time
    FROM user_actions
    WHERE order_id NOT IN (
        SELECT order_id
        FROM user_actions
        WHERE action = 'cancel_order'
    )
)
SELECT
    user_id,
    ROUND(AVG(EXTRACT(epoch FROM (time - prev_time)) / 3600))::integer AS hours_between_orders
FROM not_canceled
WHERE prev_time IS NOT NULL
GROUP BY user_id
ORDER BY user_id
LIMIT 1000
```

```python
WITH type_order AS (
  SELECT
    time :: date AS date,
    order_type,
    count(order_id) AS orders_count
  FROM (
    SELECT
      user_id,
      order_id,
      time,
      CASE WHEN time = min(time) OVER (PARTITION BY user_id) THEN 'Первый' ELSE 'Повторный' END AS order_type
    FROM user_actions
    WHERE order_id NOT IN (
      SELECT
        order_id
      FROM user_actions
      WHERE action = 'cancel_order'
    )
  ) AS orders_with_type
  GROUP BY date, order_type
  ORDER BY date, order_type
)
SELECT
  date,
  order_type,
  orders_count,
  round(
    orders_count :: numeric / sum(orders_count) OVER (PARTITION BY date)
  , 2) AS orders_share
FROM type_order
```

Отберите топ 10% курьеров по количеству доставленных за всё время заказов.У курьера, доставившего наибольшее число заказов, порядковый номер должен быть равен 1, а у курьера с наименьшим числом заказов — числу, равному десяти процентам от общего количества курьеров

```python
WITH courier_stats AS (
    SELECT
        courier_id,
        COUNT(*) filter (where action = 'deliver_order') AS orders_count
    FROM courier_actions
    GROUP BY courier_id
),
ranked AS (
    SELECT
        courier_id,
        orders_count,
        ROW_NUMBER() OVER (
            ORDER BY orders_count DESC, courier_id ASC
        ) AS courier_rank,
        COUNT(*) OVER () AS total_couriers
    FROM courier_stats
)
SELECT
    courier_id,
    orders_count,
    courier_rank
FROM ranked
WHERE courier_rank <= ROUND(0.1 * total_couriers)
ORDER BY courier_rank
```

```python
WITH courier_metrics AS (
    SELECT
        courier_id,
        (
            MAX(time) OVER ()::DATE - MIN(time) OVER (PARTITION BY courier_id)::DATE
        ) AS days_employed,
        COUNT(order_id) FILTER ( WHERE action = 'deliver_order') 
                        OVER (PARTITION BY courier_id) AS delivered_orders
    FROM courier_actions
)
SELECT DISTINCT
    courier_id,
    days_employed,
    delivered_orders
FROM courier_metrics
WHERE days_employed >= 10
ORDER BY days_employed DESC,  courier_id ASC
```

Задача 16

```python
WITH unnested_orders AS (
    SELECT
        order_id,
        creation_time,
        UNNEST(product_ids) AS product_id
    FROM orders 
),
order_price AS (
    SELECT
        uo.order_id,
        uo.creation_time,
        SUM(p.price) AS order_price
    FROM unnested_orders AS uo
    JOIN products AS p ON p.product_id = uo.product_id
    JOIN courier_actions as ca on ca.order_id = uo.order_id
    where ca.action = 'deliver_order'
    GROUP BY uo.order_id, uo.creation_time
),
daily_revenue AS (
    SELECT
        creation_time::DATE AS order_date,
        SUM(order_price) AS daily_revenue
    FROM order_price
    GROUP BY creation_time::DATE
)
SELECT
    op.order_id,
    op.creation_time,
    op.order_price,
    dr.daily_revenue,
    ROUND(
        100.0 * op.order_price / dr.daily_revenue,
        3
    ) AS percentage_of_daily_revenue
FROM order_price AS op
JOIN daily_revenue AS dr
    ON op.creation_time::DATE = dr.order_date
ORDER BY op.creation_time::DATE DESC, percentage_of_daily_revenue DESC, op.order_id ASC
```

Задача 17

Рассчитайте ежедневную выручку сервиса и отразите её. Затем посчитайте ежедневный прирост выручки. Прирост выручки отразите как в абсолютных значениях, так и в % относительно предыдущего дня.

```python
WITH canceled_orders AS (
    SELECT DISTINCT order_id
    FROM user_actions
    WHERE action = 'cancel_order'
),
order_products AS (
    SELECT
        o.order_id,
        DATE(o.creation_time) AS sales_date,
        product_id
    FROM orders AS o
    CROSS JOIN LATERAL UNNEST(o.product_ids) AS product_id
    WHERE NOT EXISTS (
        SELECT 1
        FROM canceled_orders AS co
        WHERE co.order_id = o.order_id
    )
),
daily_revenue AS (
    SELECT
        op.sales_date,
        SUM(p.price) AS revenue
    FROM order_products AS op
    LEFT JOIN products AS p
        ON p.product_id = op.product_id
    GROUP BY op.sales_date
),
revenue_with_previous_day AS (
    SELECT
        sales_date,
        revenue,
        LAG(revenue) OVER (ORDER BY sales_date) AS previous_day_revenue
    FROM daily_revenue
)

SELECT
    sales_date AS date,
    ROUND(revenue, 1) AS daily_revenue,
    ROUND(
        COALESCE(revenue - previous_day_revenue, 0),
        1
    ) AS revenue_growth_absolute,
    ROUND(
        COALESCE( (revenue - previous_day_revenue) / NULLIF(previous_day_revenue, 0) * 100, 0 ),
        1
    ) AS revenue_growth_percentage
FROM revenue_with_previous_day
ORDER BY sales_date
```

Рассчитайте медианную стоимость всех заказов. Встроенные функции для расчёта квантилей применять нельзя.

```python
WITH order_products AS (
    SELECT 
        order_id,
        UNNEST(product_ids) AS product_id
    FROM orders
    WHERE order_id NOT IN (
        SELECT order_id 
        FROM user_actions 
        WHERE action = 'cancel_order'
    )
),
order_prices AS (
    SELECT 
        o.order_id,
        SUM(p.price) AS order_price
    FROM order_products o
    LEFT JOIN products p USING(product_id)
    GROUP BY o.order_id
),
ranked_orders AS (
    SELECT 
        order_price,
        ROW_NUMBER() OVER (ORDER BY order_price) AS row_num,
        COUNT(*) OVER () AS total_rows
    FROM order_prices
)
SELECT 
    AVG(order_price) AS median_price
FROM ranked_orders
WHERE row_num BETWEEN total_rows / 2.0 AND total_rows / 2.0 + 1
```
